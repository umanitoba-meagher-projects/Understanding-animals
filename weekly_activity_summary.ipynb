# Weekly Activity Summary

This notebook provides an overview of all commits, edits, and adjustments made in the past 7 days.

```python
# Import Required Libraries
from datetime import datetime, timedelta
import subprocess
```

## Generate Timestamped Notes

This section allows you to add timestamped notes for your activities.

```python
# Function to add a timestamped log entry
def add_log_entry(note):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] {note}"
    return log_entry

# Example usage
note = "Started summarizing weekly activities."
log_entry = add_log_entry(note)
print(log_entry)
```

## Retrieve Recent Commits

This section retrieves the recent commits from the repository.

```python
# Function to get recent commits
def get_recent_commits(days=7):
    since_date = (datetime.now() - timedelta(days=days)).strftime("%Y-%m-%d")
    result = subprocess.run(
        ["git", "log", f"--since={since_date}", "--pretty=format:%h %ad | %s%d [%an]", "--date=short"],
        capture_output=True, text=True
    )
    if result.stderr:
        print("Error in get_recent_commits:", result.stderr)
    return result.stdout

# Example usage
recent_commits = get_recent_commits()
print("Recent Commits Output:", recent_commits)
```

## Retrieve Commit Details

This section retrieves detailed information about recent commits.

```python
# Function to get commit details
def get_commit_details(days=7):
    since_date = (datetime.now() - timedelta(days=days)).strftime("%Y-%m-%d")
    result = subprocess.run(
        ["git", "log", f"--since={since_date}", "--stat", "--pretty=format:%h|%ad|%an|%s", "--date=short"],
        capture_output=True, text=True
    )
    if result.stderr:
        print("Error in get_commit_details:", result.stderr)
    return result.stdout

# Example usage
commit_details = get_commit_details()
print("Commit Details Output:", commit_details)
```

## Retrieve Commit Diffs

This section retrieves the diffs of recent commits.

```python
# Function to get commit diffs
def get_commit_diffs(days=7):
    since_date = (datetime.now() - timedelta(days=days)).strftime("%Y-%m-%d")
    result = subprocess.run(
        ["git", "log", f"--since={since_date}", "-p", "--pretty=format:%h|%ad|%an|%s", "--date=short"],
        capture_output=True, text=True
    )
    if result.stderr:
        print("Error in get_commit_diffs:", result.stderr)
    return result.stdout

# Example usage
commit_diffs = get_commit_diffs()
print("Commit Diffs Output:", commit_diffs)
```

## Summarize Commit History

This section summarizes the commit history.

```python
# Function to summarize commits locally without OpenAI
from collections import Counter

def summarize_commits_locally(commit_text):
    lines = commit_text.splitlines()
    keywords = Counter()

    for line in lines:
        words = line.split()
        keywords.update(words)

    # Extract the most common keywords and their counts
    common_keywords = keywords.most_common(10)
    summary = "\n".join([f"{word}: {count}" for word, count in common_keywords])

    return f"Summary of Commit History:\n{summary}"

# Example usage
summary = summarize_commits_locally(commit_details)
print(summary)
```

## Showcase Weekly Activity

This section compiles all the information into a weekly activity summary.

```python
# Compile Weekly Activity Summary without OpenAI

def compile_weekly_summary():
    commits = get_recent_commits()
    details = get_commit_details()
    diffs = get_commit_diffs()
    summary = summarize_commits_locally(details)

    print("Weekly Activity Summary:")
    print("\nRecent Commits:\n", commits)
    print("\nCommit Details:\n", details)
    print("\nCommit Diffs:\n", diffs)
    print("\nSummary:\n", summary)

# Example usage
compile_weekly_summary()
```
